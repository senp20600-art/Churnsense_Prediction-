# Business Insights

  - **Month-to-Month contracts are the single biggest churn driver**

Customers on month-to-month contracts churn at **42.7%** versus **11.3%** for one-year and just **2.8%** for two-year contracts.Provoking the customers to take long-term contracts rather than small-term by giving some discounts for annual commitment will be beneficial.

- **Fiber optic Internet customers churn 2.2x more than DSL customers**

Despite being a premium/higher-revenue product,Fiber optic subscribers churn **41.9%** compared to **19.0%** for DSL and **7.4%** for customers with no internet service. This points possible service quality,pricing, or competitive pressure issues specific to the fiber offering that ment investigation.

- **Electronic Check payments churn 2.3x more than mailed check payments** 

Customers using electronic check for payments churn **45.3%** compared to **19.1%** for mailed check payments due to fast process and lower cost and lower security risk floowed by **16.7%** for bank-transfer payments and **15.2%** for credit-card payments.*Incentivize Auto-Pay Enrollment* by offering a one-one time bill credit or a minor discount for migraters will easily outweigh the upfront incentive cost. *Investigating the Electronic Check Journey* and *rethinking about the onboarding option* are very helpful for the growth. 

- **Lack of add-on-services(Tech Support,Online Security) strongly correlates with churn**

Customers without **TechSupport**(41.6% churn) or **OnlineSecurity**(41.8% Churn) churn nearly 3x more than those with these services (~ 15%).Bundling or proactively providing these add-ons-especially to new uprotected customers-could materially reduce churn.

- **Lack of add-on-services(Device Protection,Online Backup) strongly correlates with churn**

Customers without **Device Protection**(39.1% churn) or **OnlineBackup**(39.9% Churn) churn nearly 1.7x more than those with these services (~ 22%).Bundling or proactively providing these add-ons-especially to new uprotected customers-could materially reduce churn.
